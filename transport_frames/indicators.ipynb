```python
def indicator_terr_service_accessibility(graph: nx.MultiDiGraph, 
                                    terr: gpd.GeoDataFrame,
                                    fuel_stations: gpd.GeoDataFrame = None, 
                                    railway_stations: gpd.GeoDataFrame = None, 
                                    ports: gpd.GeoDataFrame = None,
                                    local_aerodrome: gpd.GeoDataFrame = None, 
                                    international_aerodrome: gpd.GeoDataFrame = None,
                                    bus_stops: gpd.GeoDataFrame = None) -> gpd.GeoDataFrame:
    
    sp = gpd.GeoDataFrame(geometry=terr.geometry.centroid, crs=terr.crs).reset_index()
    return indicator_service_accessibility(sp, graph, terr,
                            fuel_stations=fuel_stations, 
                            local_aerodrome=local_aerodrome,
                            international_aerodrome=international_aerodrome,
                            railway_stations=railway_stations,
                            bus_stops=bus_stops)


indicator_terr_service_accessibility(g.graph, terr,
                            fuel_stations=fuel_stations, 
                            local_aerodrome=local_aerodrome,
                            international_aerodrome=international_aerodrome,
                            railway_stations=railway_stations,
                            bus_stops=bus_stops)
```
